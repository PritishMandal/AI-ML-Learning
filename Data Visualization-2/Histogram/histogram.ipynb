```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

scores = np.random.normal(70, 10, 100)
bins = [0,30,70,90,100]

plt.hist(scores, color="skyblue", edgecolor="black", bins=bins)
plt.ylabel("# of students")
plt.xlabel("score")
print(scores)
```

```python
marks = [
   25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
   91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
   92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

plt.hist(marks, color="skyblue", edgecolor="black", bins=bins)
plt.ylabel("# of students")
plt.xlabel("scores")
plt.title("distribution of marks for students")
```

```python
 # Multiple datasets
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

plt.hist(fraud_transactions, bins=10, label='Fraud', color='red', alpha=0.5, edgecolor="black")
plt.hist(legit_transactions, bins=10, label='Legit', color='green', alpha=0.5, edgecolor="black")

plt.xlabel('Transaction Amount ($)')
plt.ylabel('Frequency')
plt.title('Transaction Amount Distribution: Legitimate vs Fraudulent')
plt.legend()
plt.grid(True)
```

```python
# Box Plot
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data)
plt.grid(True)
```

```python
# Multiple Datasets

group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)

plt.boxplot([group1, group2], tick_labels=["Group1", "Group2"])
plt.show()
```

```python
# Horizontal Box Plot
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.boxplot(data, vert=False, showmeans=True, whis=2.0)
plt.grid(True)
```
